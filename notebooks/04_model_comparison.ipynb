# Model Comparison

This notebook compares the Standard CNN and Lightweight CNN.

The comparison focuses on:

- Accuracy
- Parameter count
- Model size
- MACs
- FLOPs
- CPU latency
- GPU latency


```python
import matplotlib.pyplot as plt

models = ["Standard CNN", "Lightweight CNN"]

accuracy = [99.05, 97.23]
parameters = [896906, 3989]
model_size = [3.4214, 0.0165]
macs = [8.255, 0.376]
flops = [16.510, 0.752]
cpu_latency = [0.613, 1.000]
gpu_latency = [0.815, 1.223]

print("Accuracy:", dict(zip(models, accuracy)))
print("Parameters:", dict(zip(models, parameters)))
print("Model size:", dict(zip(models, model_size)))
print("MACs:", dict(zip(models, macs)))
print("FLOPs:", dict(zip(models, flops)))
print("CPU latency:", dict(zip(models, cpu_latency)))
print("GPU latency:", dict(zip(models, gpu_latency)))
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(models, accuracy)
ax.set_ylabel("Accuracy (%)")
ax.set_title("Test Accuracy Comparison")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(models, parameters)
ax.set_ylabel("Trainable Parameters")
ax.set_title("Parameter Count Comparison")
ax.set_yscale("log")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))

ax.bar(models, macs)
ax.set_ylabel("MACs (Millions)")
ax.set_title("Computational Complexity")

plt.tight_layout()
plt.show()
```

```python
print("Parameter reduction:",
      parameters[0] / parameters[1])

print("Model-size reduction:",
      model_size[0] / model_size[1])

print("MAC reduction:",
      macs[0] / macs[1])

print("FLOP reduction:",
      flops[0] / flops[1])

print("Accuracy difference:",
      accuracy[0] - accuracy[1],
      "percentage points")
```

## Final Research Observation

The Lightweight CNN reduces model parameters, model size, MACs, and FLOPs dramatically.

However, the benchmark shows that lower theoretical computational complexity does not automatically produce lower single-image latency on general-purpose CPU/GPU hardware.

This demonstrates the importance of evaluating both theoretical complexity and real hardware performance.
